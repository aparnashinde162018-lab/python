```python
from selenium import webdriver
from bs4 import BeautifulSoup
```

```python
drive=webdriver.Chrome()
```

```python
drive.get("https://www.amazon.in/s?k=mobile&rh=n%3A1389401031&ref=nb_sb_noss")
```

```python
drive.page_source
```

```python
data=BeautifulSoup(drive.page_source,"html.parser")
data
print(data.prettify())
```

```python
productname=data.find_all("span",{"class":"a-size-medium a-color-base a-text-normal"})
product=[p.text for p in productname]
product=product[0:23]
product
```

```python
ratings=data.find_all("span",{"class":"a-icon-alt"})
RATING=[r.text for r in ratings]
RATING=RATING[0:23]
RATING
```

```python
reviews=data.find_all("span",{"class":"a-size-base s-underline-text"})
REVIEWS=[i.text for i in reviews]
REVIEWS
```

```python
price=data.find_all("span",{"class":"a-price-whole"})
PRICE=[j.text for j in price]
PRICE=PRICE[0:23]
PRICE
```

```python
print(len(product))
print(len(RATING))
print(len(REVIEWS))
print(len(PRICE))
```

```python
import pandas as pd
```

```python
df=pd.DataFrame({"TITLE":product,"RATINGS":RATING,"REVIEWS":REVIEWS,"PRICE":PRICE})
df
```

```python
df.to_csv("MYAMAZON.csv",index=False)
```
